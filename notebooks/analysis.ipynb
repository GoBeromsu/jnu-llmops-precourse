```python
import sys, pathlib
# 이 Notebook 은 저장소 Root 에서 실행한다고 가정한다. notebooks/ 안에서 열려도 Root 를 찾도록 한 줄 보정.
ROOT = pathlib.Path.cwd() if (pathlib.Path.cwd() / 'records.py').exists() else pathlib.Path.cwd().parent
sys.path.insert(0, str(ROOT))
print('저장소 Root:', ROOT)
```

```python
import json
with open(ROOT / 'experiment_records.json', encoding='utf-8') as f:
    RECORDS = json.load(f)
len(RECORDS)
```

```python
# >>> TODO-GUIDED: D4-IMPORT >>>
# Day 4: 아래에 복사돼 있는 검색 Loop 를 지우고, records.py 의 find_records 를 import 해서 부른다.
def find_warnings(records, equipment_id):
    found = []
    for record in records:
        if record['equipment_id'] != equipment_id:
            continue
        if record['status'] != 'Warning':
            continue
        found.append(record)
    return found
# <<< TODO-GUIDED: D4-IMPORT <<<
```

```python
EQUIPMENT = ['HPLC-01', 'GC-02', 'CENT-03']
counts = {e: len(find_warnings(RECORDS, e)) for e in EQUIPMENT}
counts  # 기대값: {'HPLC-01': 2, 'GC-02': 0, 'CENT-03': 0}
```
